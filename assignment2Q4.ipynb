```python
import pandas as pd
```

```python
titanic = pd.read_csv("titanic_cleaned.csv")
fares = pd.read_csv("ticket_fares.csv")
```

```python
df = pd.merge(titanic, fares, on='Ticket', how='left')
```

```python
bins = [0, 12, 19, 59, 120]
labels = ['Child', 'Teen', 'Adult', 'Senior']

df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)

survivalAgeSex = df.groupby(['Sex', 'AgeGroup'])['Survived'].mean()

print('Mean Survival by Sex & AgeGroup')
print(survivalAgeSex)
print("\n")
```

```python
survivalByClass = df.groupby('Pclass')['Survived'].mean()

print('Mean Survival by Class (Method A)')
print(survivalByClass)
print("\n")
```

```python
df['FareBin'] = pd.qcut(df['Fare_y'], q=4, labels=['Low', 'Medium', 'High', 'VeryHigh'])

survivalByFarebin = df.groupby('FareBin')['Survived'].mean()

print('Mean Survival by Fare Bin (Method B)')
print(survivalByFarebin)
print("\n")
```

```python
with open("report.txt", "w") as f:

    f.write("Hypothesis 1: Women and Children First\n")
    f.write('****************************************\n')
    f.write("The mean survival rates grouped by Sex and AgeGroup suggest the\n")
    f.write("following conclusions:\n\n")
    f.write(str(survivalAgeSex))
    f.write("\n\nBased on the data, women show a significantly higher survival rate\n")
    f.write("than men, and younger passengers (children) also show higher\n")
    f.write("survival values compared to adults. Therefore, the hypothesis\n")
    f.write("'women and children first' is supported by the data.\n\n\n")

    f.write("Hypothesis 2: Wealthy Had Higher Survival Rates\n")
    f.write("***************************************************\n")
    f.write("Method A (Class-based survival):\n")
    f.write(str(survivalByClass))
    f.write("\n\nMethod B (Fare-based quantile groups):\n")
    f.write(str(survivalByFarebin))
    f.write("\n\nBoth methods show that passengers paying higher fares or belonging\n")
    f.write("to higher classes had a better chance of survival. Thus, the\n")
    f.write("hypothesis that 'the wealthy survived more' is supported.\n")

print("Report written successfully to report.txt")
```
