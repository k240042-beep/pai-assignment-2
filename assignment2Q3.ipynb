```python
import pandas as pd
```

```python
df=pd.read_csv('Titanic-Dataset.csv')
```

```python
reportLines=[]
reportLines.append('column inspection report\n')
reportLines.append('************************\n')
for col in df.columns:
  dataType=df[col].dtype
  missingP=df[col].isnull().mean()*100
  reportLines.append(f'column:{col}')
  reportLines.append(f'Data Type:{dataType}')
  reportLines.append(f'missing % :{missingP}')
```

```python
with open('inspect_report.txt','w') as f:
  f.writelines(reportLines)
print('report is saved as inspection_report.txt')
```

```python
ageGroupMedians = df.groupby(['Pclass', 'Sex'])['Age'].transform('median')
df['Age'] = df['Age'].fillna(ageGroupMedians)
```

```python
embarkedMode=df['Embarked'].mode()[0]
df['Embarked']=df['Embarked'].fillna(embarkedMode)
```

```python
df = df.drop(columns=['Cabin'])
```

```python
df['FamilySize']=df['SibSp']+df['Parch']
df['IsAlone']=(df['FamilySize']==0).astype(int)
```

```python
df['Age']=df['Age'].astype('int64')
```

```python
df.to_csv('Titani_cleaned.csv',index=False)
print('Clean Dta saved as Titanic_cleaned.csv')
```
