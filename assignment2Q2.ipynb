```python
import numpy as np
filename = "sensor_data.csv"

nestedList = []
with open(filename, "r") as f:
    for line in f:
        parts = line.strip().split(",")
        row = []
        for item in parts:
            try:
                row.append(float(item))
            except ValueError:
                row.append(-999.0)
        nestedList.append(row)

data = np.array(nestedList, dtype=float)

data[data == -999.0] = np.nan

data[data < 0] = np.nan
data[data > 100] = np.nan

sensorMeans = np.nanmean(data, axis=0)

hourMedians = np.nanmedian(data, axis=1)

nanCounts = np.isnan(data).sum(axis=0)
worstSensorIndex = np.argmax(nanCounts)

print("DATA ANALYSIS RESULTS:")
print("Mean moisture for each sensor (100 sensors):")
print(sensorMeans)

print("\nMedian moisture for each hour (720 hours):")
print(hourMedians)

print(f"\nSensor with highest invalid readings: Sensor #{worstSensorIndex}")
print(f"Invalid reading count: {nanCounts[worstSensorIndex]}")

minVal = np.nanmin(data)
maxVal = np.nanmax(data)

normalized = (data - minVal) / (maxVal - minVal)

output_file = "sensor_data_normalized.csv"
np.savetxt(output_file, normalized, delimiter=",")

print(f"\nNormalized dataset saved as: {output_file}")
```
